```python
# All imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
# Load inference data
path = os.getcwd()
data_path  = '/robotics_backup/results_week3_sweep.csv'
full_data_path = path + data_path

pd_ablation = pd.read_csv(full_data_path)
pd_ablation.head(10)
```

```python
# Group data by perturbation and severity

# Group by the two categories and calculate the average for specified columns
result_df = pd_ablation.groupby(['policy','perturb_type', 'severity'])[[
    'ep_return',
    'ep_len',
    'track_err',
    'fell',
    'hit_ceiling',
]].mean().reset_index()

print('Averaged result over 30 different robots initiated')
print(result_df.head(10))

print('Number of unique perturbation & severity combinations')
unique_combinations_count = pd_ablation.groupby(['perturb_type', 'severity']).ngroups
print(unique_combinations_count)

print('List all perturbation & severity combinations:')
all_pairs = pd_ablation[['perturb_type', 'severity']].drop_duplicates().values.tolist()
print(all_pairs)
```

```python
# Plot heatmaps:

# ── define the canonical orders for the plot ──────────────────
ROW_ORDER = [
    "nominal + none",
    "friction + moderate",       "friction + hard",
    "mass + moderate",           "mass + hard",
    "motor_strength + moderate", "motor_strength + hard",
    "sensor_noise + moderate",   "sensor_noise + hard",
    "action_latency + moderate", "action_latency + hard",
]

COL_ORDER = [
    "baseline_full_dr",
    "no_friction",
    "no_mass",
    "no_motor_strength",
    "no_sensor_noise",
    "no_action_latency",
]

# Choose parameter to evaluate policy on:
parameters = ['ep_return', 'ep_len', 'track_err', 'fell']#,'hit_ceiling']

for parameter in parameters:

    # Create pandas dataframe for this particular parameter grouped by policy, perturb_type and severity:
    pd_mean = pd_ablation.groupby(['policy', 'perturb_type', 'severity'])[parameter].mean().reset_index()
    
    # Create combined Y-axis label column:
    pd_mean['perturb_severity'] = pd_mean['perturb_type'] + " + " + pd_mean['severity']
    
    # Pivot dataframe into matrix suitable for heatmap:
    heatmap_data = pd_mean.pivot(
        index = 'perturb_severity',
        columns = 'policy',
        values = parameter
    )

    heatmap_data = heatmap_data.reindex(index = ROW_ORDER, columns = COL_ORDER)

    if parameter == 'fell':
        normalized_heatmap_data = heatmap_data
    else:
        baseline_values = heatmap_data['baseline_full_dr'].values
        # Divide the entire matrix by the baseline row:
        normalized_heatmap_data = heatmap_data.div(baseline_values, axis=0)
    
    # Make a plot:
    plt.figure(figsize = (10,8))
    sns.heatmap(
        normalized_heatmap_data,
        annot = True,
        center=1.0, 
        cmap='RdYlGn', # cmap = 'viridis'
        fmt = '.2f',
        cbar_kws = {'label': parameter}
    )
                                
    plt.title('Performance Across Policy and Perturbation Combinations')
    plt.ylabel('Perturb Type + Severity')
    plt.xlabel('Policy')
    plt.tight_layout()
    plt.show()
```

```python
# Sanity checks & improved plotting with some labels:

# ═══ SANITY CHECKS (run once, before anything) ═══════════════════════
assert pd_ablation.groupby('policy').size().eq(330).all(), "missing episodes!"
assert pd_ablation['fell'].dtype == bool, "fell not parsed as bool!"
print(pd_ablation.dtypes)

# ═══ HEATMAPS ════════════════════════════════════════════════════════
# metric -> (normalize?, colormap, direction note)
METRICS = {
    'ep_return':  (True,  'RdYlGn',   'ratio vs baseline — ↑ better'),
    'track_err':  (True,  'RdYlGn_r', 'ratio vs baseline — ↓ better'),   # _r flips: high ratio = red
    'fell':       (False, 'Reds',     'fall rate — ↓ better'),
    'ep_len':     (True,  'RdYlGn',   'ratio vs baseline — ↑ better (caveat: ambiguous under latency)'),
}

for parameter, (normalize, cmap, note) in METRICS.items():
    pd_mean = pd_ablation.groupby(['policy', 'perturb_type', 'severity'])[parameter].mean().reset_index()
    pd_mean['perturb_severity'] = pd_mean['perturb_type'] + " + " + pd_mean['severity']

    heatmap_data = pd_mean.pivot(index='perturb_severity', columns='policy', values=parameter)
    heatmap_data = heatmap_data.reindex(index=ROW_ORDER, columns=COL_ORDER)
    assert not heatmap_data.isna().any().any(), f"NaN stripe in {parameter} — label mismatch!"

    if normalize:
        plot_data = heatmap_data.div(heatmap_data['baseline_full_dr'], axis=0)
        center, vmin, vmax = 1.0, None, None
    else:
        plot_data = heatmap_data
        center, vmin, vmax = None, 0, 1

    plt.figure(figsize=(10, 8))
    sns.heatmap(plot_data, annot=True, fmt='.2f', cmap=cmap,
                center=center, vmin=vmin, vmax=vmax,
                cbar_kws={'label': note})
    plt.title(f'{parameter} — {note}')
    plt.ylabel('Perturbation + Severity'); plt.xlabel('Policy')
    plt.tight_layout()
    plt.savefig(f'heatmap_{parameter}.png', dpi=150, bbox_inches='tight')
    plt.show()
```

```python
# ═══ STEP 4: ERROR BARS ══════════════════════════════════════════════
parameter = 'fell'#'ep_return' 'ep_len', 'track_err', 'fell'
stats = (pd_ablation
         .groupby(['policy', 'perturb_type', 'severity'])[parameter]
         .agg(['mean', 'std', 'count']))
stats['sem'] = stats['std'] / np.sqrt(stats['count'])

# Is each policy's gap vs baseline bigger than the combined noise? (~2-sigma test)
base = stats.xs('baseline_full_dr', level='policy')
gaps = stats.join(base, on=['perturb_type', 'severity'], rsuffix='_base')
gaps['gap']        = gaps['mean'] - gaps['mean_base']
gaps['gap_sigma']  = gaps['gap'].abs() / np.sqrt(gaps['sem']**2 + gaps['sem_base']**2)
gaps['significant'] = gaps['gap_sigma'] > 2

print(gaps[gaps.index.get_level_values('policy') != 'baseline_full_dr']
      [['gap', 'gap_sigma', 'significant']]
      .sort_values('gap_sigma', ascending=False)
      .round(2)
      .to_string())
```

```python
# Use this siginificance metric on the plots themselves:

def sig_matrix(df, parameter, binomial=False):
    """Boolean matrix (perturb_severity x policy): |gap vs baseline| > 2 sigma."""
    stats = df.groupby(['policy', 'perturb_type', 'severity'])[parameter].agg(['mean', 'std', 'count'])
    if binomial:   # for 'fell': proportion, not continuous mean
        p = stats['mean']
        stats['sem'] = np.sqrt(p * (1 - p) / stats['count'])
    else:
        stats['sem'] = stats['std'] / np.sqrt(stats['count'])
    base = stats.xs('baseline_full_dr', level='policy')
    gaps = stats.join(base, on=['perturb_type', 'severity'], rsuffix='_base')
    gap_sigma = (gaps['mean'] - gaps['mean_base']).abs() / np.sqrt(gaps['sem']**2 + gaps['sem_base']**2)
    out = (gap_sigma > 2).reset_index()
    out['perturb_severity'] = out['perturb_type'] + " + " + out['severity']
    return (out.pivot(index='perturb_severity', columns='policy', values=0)
              .reindex(index=ROW_ORDER, columns=COL_ORDER)
              .fillna(False))

# ═══ HEATMAPS ════════════════════════════════════════════════════════
# metric -> (normalize?, colormap, direction note)
METRICS = {
    'ep_return':  (True,  'RdYlGn',   'ratio vs baseline — ↑ better'),
    'track_err':  (True,  'RdYlGn_r', 'ratio vs baseline — ↓ better'),   # _r flips: high ratio = red
    'fell':       (False, 'Reds',     'fall rate — ↓ better'),
    'ep_len':     (True,  'RdYlGn',   'ratio vs baseline — ↑ better (caveat: ambiguous under latency)'),
}

for parameter, (normalize, cmap, note) in METRICS.items():
    pd_mean = pd_ablation.groupby(['policy', 'perturb_type', 'severity'])[parameter].mean().reset_index()
    pd_mean['perturb_severity'] = pd_mean['perturb_type'] + " + " + pd_mean['severity']

    heatmap_data = pd_mean.pivot(index='perturb_severity', columns='policy', values=parameter)
    heatmap_data = heatmap_data.reindex(index=ROW_ORDER, columns=COL_ORDER)
    assert not heatmap_data.isna().any().any(), f"NaN stripe in {parameter} — label mismatch!"

    if normalize:
        plot_data = heatmap_data.div(heatmap_data['baseline_full_dr'], axis=0)
        center, vmin, vmax = 1.0, None, None
    else:
        plot_data = heatmap_data
        center, vmin, vmax = None, 0, 1

    plt.figure(figsize=(10, 8))
    ax = sns.heatmap(plot_data, annot=True, fmt='.2f', cmap=cmap,
                center=center, vmin=vmin, vmax=vmax,
                cbar_kws={'label': note})
    sig = sig_matrix(pd_ablation, parameter, binomial=(parameter == 'fell'))
    for i in range(sig.shape[0]):
        for j in range(sig.shape[1]):
            if sig.iloc[i, j]:
                ax.add_patch(plt.Rectangle((j, i), 1, 1, fill=False, edgecolor='red', lw=2.5))
    plt.title(f'{parameter} — {note}')
    plt.ylabel('Perturbation + Severity'); plt.xlabel('Policy')
    plt.tight_layout()
    plt.savefig(f'heatmap_{parameter}.png', dpi=150, bbox_inches='tight')
    plt.show()
```
